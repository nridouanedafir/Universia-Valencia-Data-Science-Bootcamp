## Ejercicio 5
En un nuevo array, sustituye todos los impares de `my_array` por -1

````python
a=np.zeros(shape=(10,1))

for i in range(0,10):
    if my_array[i]%2==0:
        a[i]=my_array[i]
    else:
        a[i]=-1
print(a)
````

## Ejercicio 6

````python
array2=np.where(my_array%2!=0,-1,my_array)
print(array2)
````
